```python
#1. ANALISIS DE VALORES FALTANTES
# A. Definimos valores con  muchos faltantes y pocos faltantes
columnas_nulas=[]
columnas_no_nulas=[]
for i in range(0,len(data.columns)):
    a=data.columns[i]
    if data[a].isnull().sum()>(0.4*len(data)):
        columnas_nulas.append(a)
    else:
        columnas_no_nulas.append(a)
print(data.shape)
# B. Eliminamos columnas con muchos nulos
data2= data.drop(columnas_nulas,axis=1)
print(data2.shape)

# C. Definimos columnas categoricas y numericas
columnas_numericas=[]
columnas_categoricas=[]
for i in range(0,len(columnas_no_nulas)):
    col=columnas_no_nulas[i]
    if is_object_dtype(data2[col]):
        columnas_categoricas.append(col)
    else:
        columnas_numericas.append(col)

# D. Remplazamos la variables catergoricas por 0 y las numericas por el promedio y la columnas "total_rev_hi_lim" por los valores de "funded_amnt"
data['total_rev_hi_lim'] = data['total_rev_hi_lim'].fillna(data["funded_amnt"])

for i in range(0,len(columnas_numericas)):
    col=columnas_numericas[i]
    data2[col]=data2[col].fillna(data[col].mean())

for i in range(0,len(columnas_categoricas)):
    col=columnas_categoricas[i]
    data2[col]=data2[col].fillna(0)
```

```python
# 2. ANALIZAMOS COLUMNA CATEGORICAS QUE DEBEN SER ENTEROS
# A.. Columnas term (para que quede en entero sacamos la palabra months)
print(data2['term'].unique())
data2['term'] = data2['term'].replace({'months': '', ' ': ''}, regex=True)
data2['term']=data2['term'].astype(int)
print(data2['term'].unique())

# B. Columnas emp_lenght (para que quede en entero sacamos la palabra yeasrs y otras)
print(data2['emp_length'].unique())
data2['emp_length'] = data2['emp_length'].replace({
    r'10\+ years': '10',
    r'< 1 year': '0',
    r'< years': '',
    r' years': '',
    r' year': ''
}, regex=True)
data2['emp_length']=data2['emp_length'].fillna("0")
data2['emp_length']=data2['emp_length'].astype(int)
print(data2['emp_length'].unique())
```

```python
# 2. ANALIZAMOS COLUMNA CATEGORICAS QUE DEBEN SER FECHAS
# A. Columna: 'issue_d'
columnas_fecha=['issue_d',"earliest_cr_line"]
columnas_nuevas=[["fecha_credito","meses_credito"],["fecha_ultimo_credito","meses_desde_ultimo_credito"]]
fecha_fija=pd.to_datetime('2022-01-01')
for i in range(0,len(columnas_fecha)):
    a=columnas_fecha[i]
    b=columnas_nuevas[i]
    c=b[0]
    d=b[1]
    print("......",d,"........")
    data2[c]=pd.to_datetime(data[a],format="%b-%y", errors='coerce')
    data2[d] = ((fecha_fija.year - data2[c].dt.year) * 12 + (fecha_fija.month - data2[c].dt.month))
    print(data2[d])
```

```python
# ONE-HOT-ENCODING
# A. Hacemos on-hot-encoding sobre variables categoricas
columnas=["home_ownership","purpose","grade","initial_list_status","addr_state","loan_status","verification_status","sub_grade"]
prefijos=["home","proposito","grado","Estad_lita","Ubicacion","Estado_credito","Estado_verificacion","Subgrado"]
datos_dummies=[]
for i in range(0,len(columnas)):
    a=columnas[i]
    b=prefijos[i]
    data_dumm=pd.get_dummies(data2[a], prefix=b,prefix_sep="_", dtype=int)
    datos_dummies.append(data_dumm)
datos_dummies=pd.concat(datos_dummies,axis=1)
print("El tamaño de data es: ",data.shape)
print("El tamaño de data2 es: ",data2.shape)
print("El tamaño de las dummies es: ",datos_dummies.shape)
data_concatenados=pd.concat([data2,datos_dummies],axis=1)
print("El tamaño de nuevo dataset es: ",data_concatenados.shape)
```

```python
# 1. FUNCION PARA EDA INICAL
# Configurar Pandas para mostrar todo
pd.set_option('display.max_rows', None)  # Muestra todas las filas
pd.set_option('display.max_columns', None)  # Muestra todas las columnas
pd.set_option('display.width', None)  # Evita el truncamiento por ancho
pd.set_option('display.max_colwidth', None)  # Muestra todo el contenido

def eda_report(df, name):
    print("\n" + "="*60)
    print(f"EDA REPORT: {name}")
    print("="*60)

    # Información básica
    print("\n--- FORMA DEL DATAFRAME ---")
    print(f"Filas: {df.shape[0]} | Columnas: {df.shape[1]}")

    print("\n--- PRIMERAS FILAS ---")
    print(df.head())

    print("\n--- TIPOS DE DATOS POR COLUMNA ---")
    print(df.dtypes)

    # Valores nulos
    print("\n--- VALORES NULOS POR COLUMNA ---")
    nulos = df.isnull().sum()
    porcentaje = (nulos / len(df)) * 100
    resumen_nulos = pd.DataFrame({
        'Cantidad de nulos': nulos,
        'Porcentaje (%)': porcentaje.round(2)
    })
    resumen_nulos = resumen_nulos[resumen_nulos['Cantidad de nulos'] > 0]
    
    if resumen_nulos.empty:
        print("No hay valores nulos.")
    else:
        print(resumen_nulos.sort_values(by='Cantidad de nulos', ascending=False))

    # Descripción de variables numéricas
    print("\n--- DESCRIPCIÓN DE VARIABLES NUMÉRICAS ---")
    numeric_cols = df.select_dtypes(include=np.number).columns
    if len(numeric_cols) > 0:
        print(df[numeric_cols].describe())
    else:
        print("No hay variables numéricas.")

    # Descripción de variables categóricas
    print("\n--- DESCRIPCIÓN DE VARIABLES CATEGÓRICAS ---")
    cat_cols = df.select_dtypes(include='object').columns
    if len(cat_cols) > 0:
        print(df[cat_cols].describe())
    else:
        print("No hay variables categóricas.")
```

```python
def PDE_categorica(df,columna,df_var_dependiente):
    # C. PESO DE EDVIDENCIA DE VARIABLES Y VALOR DE LA INFORMACION
    df1=pd.concat([df[columna], df_var_dependiente], axis=1)
    data_a=df1.groupby(df1.columns.values[0],as_index=False)[df1.columns.values[1]].count() # cantidad instancias buenas por categoria
    data_b=df1.groupby(df1.columns.values[0],as_index=False)[df1.columns.values[1]].mean()  #promedio instancias buenas  
    data_c=pd.concat([data_a,data_b],axis=1)
    data_c=data_c.iloc[:,[0,1,3]]  # nueva tabla de evidencia
    data_c.columns=[data_c.columns[0],"instancias","%_buenos"]
    data_c["%_instancias"]=data_c["instancias"]/data_c["instancias"].sum() # porcentaje de instancias buenas por categroia
    data_c["n_buenos"]=data_c["%_buenos"]*data_c["instancias"]  #cantidad de instancias buenas por categroia
    data_c["n_malos"]=(1-data_c["%_buenos"])*data_c["instancias"] #cantidad de instancias malas por categroia
    data_c["prop_%_buenos"]=data_c["n_buenos"]/data_c["n_buenos"].sum() # porcentaje de buenos en el total por categroia
    data_c["prop_%_malos"]=data_c["n_malos"]/data_c["n_malos"].sum() # porcentaje de malos en el total por categroia
    data_c["PDE"]=np.log(data_c["prop_%_buenos"]/data_c["prop_%_malos"]) # Peso de la Evidencia= logaritimo de %Buenos/%Malos
    data_c=data_c.sort_values(["PDE"]).reset_index(drop=True)
    data_c["delta_prop_%_buenos"]=data_c["prop_%_buenos"].diff().abs()
    data_c["delta_PDE"]=data_c["PDE"].diff().abs()
    
    #  b. Sobre la variabl  "home_ownership": Valor de la Informacion
    data_c["VI"]=data_c["PDE"]*(data_c["prop_%_buenos"]- data_c["prop_%_malos"])
    data_c.replace([np.inf,-np.inf],np.nan,inplace=True) #quito el valor infinito
    data_c["SUMATORIA VI"]= data_c1["VI"].sum() 
    return data_c
```
